```python
import pandas as pd

# 1. Define the dataset matching the image
data = [
    ['Big', 'Red', 'Circle', 'No'],
    ['Small', 'Red', 'Triangle', 'No'],
    ['Small', 'Red', 'Circle', 'Yes'],
    ['Big', 'Blue', 'Circle', 'No'],
    ['Small', 'Blue', 'Circle', 'Yes']
]

# Create a DataFrame for visual clarity
columns = ['Size', 'Color', 'Shape', 'Class']
df = pd.DataFrame(data, columns=columns)

print("--- Training Dataset ---")
print(df)
print("\n" + "="*40 + "\n")

# 2. Implement FIND-S Algorithm
def train_find_s(df):
    # Extract features (X) and targets (y)
    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    num_features = X.shape[1]

    # Initialize hypothesis with null values (ϕ)
    hypothesis = ['ϕ'] * num_features
    print(f"Initial Hypothesis: {hypothesis}\n")

    # Trace step-by-step updates
    for i, (instance, label) in enumerate(zip(X, y)):
        if label.strip().lower() == 'yes':
            # If hypothesis is uninitialized, set it to the first positive instance
            if hypothesis == ['ϕ'] * num_features:
                hypothesis = list(instance)
            else:
                # Compare features and generalize to '?' if they differ
                for j in range(num_features):
                    if hypothesis[j] != instance[j]:
                        hypothesis[j] = '?'
            print(f"Row {i+1} (Positive):  Updated Hypothesis -> {hypothesis}")
        else:
            print(f"Row {i+1} (Negative):  Ignored -> Hypothesis remains {hypothesis}")

    return hypothesis

# 3. Run the algorithm and print final output
final_hypothesis = train_find_s(df)

print("\n" + "="*40)
print(f"Most Specific Hypothesis: {final_hypothesis}")
```
